```python
import requests
import pandas as pd
from datetime import datetime
```

```python
data_atual = datetime.now().strftime("%m-%d-%Y")
url_api = f"https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/CotacaoDolarPeriodo(dataInicial=@dataInicial,dataFinalCotacao=@dataFinalCotacao)?@dataInicial='01-01-2026'&@dataFinalCotacao='{data_atual}'&$top=500&$format=json&$select=cotacaoCompra,cotacaoVenda,dataHoraCotacao"
dados_api = requests.get(url_api)

try:
    dados_api.raise_for_status()  # Verifica se a requisição foi bem-sucedida

    dados_json = dados_api.json()
    dados_df = pd.DataFrame(dados_json['value'])

    tabela_bronze = dados_df.to_csv('tabela_bronze.csv', index=False)
    
except requests.exceptions.HTTPError as http_err:
    print(f"Ocorreu um erro HTTP: {http_err}")
except Exception as err:
    print(f"Ocorreu um erro: {err}")
```
